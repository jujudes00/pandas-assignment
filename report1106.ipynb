### 5. 복합 조건

각각 조건을 만족하는 행 수를 출력하세요.

1. 3등석 **그리고** 남성 **그리고** 생존한 승객 (`&`)
2. 1등석 **또는** 요금 100 이상인 승객 (`|`)
3. 혼자 탑승하지 **않은** 승객 (`~`, `alone` 열 사용)

```python
mask = (titanic["class"] == "Third") & (titanic["sex"] == "male") & (titanic["alive"] == "yes")
mask2 = (titanic["class"] == "First") | (titanic["fare"] >= 100)
mask3 = (titanic["alone"] == True)
print(titanic[mask].shape[0])
print(titanic[mask2].shape[0])
print(titanic[~mask3].shape[0])
```

3. 같은 범위를 양 끝 미포함(`inclusive="neither"`)으로 바꿔 `shape`를 출력하세요.

### 7. loc[조건, 열]

### 10. 종합 — 20~40세 남성, 1·2등석 vs 3등석

질문: "20~40세 남성 승객의 생존률은 객실 등급에 따라 얼마나 다를까?"

1. 남성 · 나이 20~40세(`between`) · 1·2등석(`isin`) 조건으로 `pclass`, `age`, `fare`, `survived` 열을 선택하세요.
2. 행 수, 생존률(소수 둘째 자리), 요금 상위 3명을 출력하세요.
3. 같은 조건을 `query`로 써서 행 수를 검증하세요.
4. 등급만 3등석으로 바꿔 생존률을 구하고, 두 생존률을 비교하는 해석을 마크다운 셀에 2~3문장으로 쓰세요.

```python
# 1. 
mask = (titanic["sex"] == "male") & (titanic["age"].between(20, 40)) & (titanic["pclass"].isin([1,2]))
x = titanic.loc[mask,["pclass", "age", "fare", "survived"]]
print(x)

# 2.
print(x.shape[0])
print(round(x["survived"].mean(), 2))
print(x.sort_values(by="fare", ascending=False).head(3))

# 3. 
y = titanic.query("sex == 'male' and age >= 20 and age <= 40 and pclass in [1, 2]")
print(x.shape[0])
print(y.shape[0])

# 4. 
mask2 = (titanic["sex"] == "male") & (titanic["age"].between(20, 40)) & (titanic["pclass"] == 3)
a = titanic[mask2]
print(round(x["survived"].mean(), 2))
print(round(a["survived"].mean(), 2))
```

* 1,2등석 생존률은 0.24이고 3등석은 0.15 결과 값이 나왔다
* 그래서 1,2등석 생존률이 3등석보다 0.9 높다
